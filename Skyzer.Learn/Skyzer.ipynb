## Import, Look at, and Clean our Data

```python
import pandas as pd

sb_filepath = "./sb_data.csv"
sb_data = pd.read_csv(sb_filepath)
print("Description")
sb_data.describe()
print("Columns")
sb_data.columns
# dropna drops missing values
sb_data = sb_data.dropna(axis=0)
```

## Set up the First Model

```python
# Prediction Target
y = sb_data.Price
# Features that we use to predict
# DyeColor only for armor??
sb_features = ["Item:SkyblockId", "Item:DisplayName", "Item:DyeColor", "Item:Reforge", "Item:Enchantments", "Item:Stars", "Item:HotPotatoCount", "Item:Recombobulated"]
X = sb_data[sb_features]
print("look at the data")
X.describe()
print("top 5 item's features")
X.head()
```

## Create First Model (decision tree regressor)

```python
from sklearn.tree import DecisionTreeRegressor

# define our model
# we set the random state so we get the same result every run
sb_model = DecisionTreeRegressor(random_state=7)

# fit the model
# fitting the model capatures patterns from the data
sb_model.fit(X, y)
```

## Try Predicting!

```python
print("Making predictions for the top 5 items")
print(X.head())
print("predictions:")
print(sb_model.predict(X.head()))
```
